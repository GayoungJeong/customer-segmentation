```python
# 셀 2: 데이터 불러오기 (시트 2개를 하나로 합침, 1~2분 걸려요)
import pandas as pd
import os

# 실제 파일이 있는 경로로 지정
path = '/content/online_retail_II.xlsx'
df = pd.concat(pd.read_excel(path, sheet_name=None), ignore_index=True)

# 저장할 드라이브 디렉토리가 없으면 생성
save_dir = '/content/drive/MyDrive/marketing-portfolio/data'
os.makedirs(save_dir, exist_ok=True)

# 다음부터 빨리 불러오게 csv로 저장
df.to_csv(os.path.join(save_dir, 'retail.csv'), index=False)

# 데이터 확인
display(df.head())
```
