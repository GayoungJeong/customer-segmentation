관찰2. 숫자열에 대한 요약
min: 가장 작은 값. 원래는 수량이나 가격이 마이너스일 수 없는데 실제 min값은 마이너스처럼 나옴.
max: 가장 큰 값인데 수량이 80000이 넘는 것임. 숫자가 현실적이지 못함

(참고: 숫자가 1.067371e+06처럼 나오면 1.067371 × 10⁶, 즉 약 107만이라는 뜻)

아래 결과를 통해  80995개를 구매했다가 12분 뒤 취소한 기록을 볼 수 있음. 즉,오류가 아님.

(i)수량이 음수인 줄은 두 종류임. ① Invoice가 C로 시작하는 취소 건 ② 가격이 0이고 고객 ID와 상품명이 없는 줄(재고 조정 기록으로 추정). 둘 다 실제 구매가 아니므로 고객 분석에서 처리 방법을 정해야 함

(ii)가격이 마이너스인 줄은 Invoice가 A로 시작하고, Description이 “Adjust bad debt”임 → 상품 판매가 아니라 대손 처리(회계 조정) 기록으로 추정. 누가 얼마나 샀느냐를 나타내는 요소가 없기 때문에 고객 분석에서는 제외해야 함.
